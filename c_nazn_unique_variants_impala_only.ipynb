# Уникальные варианты `c_nazn` (Impala)

Минимальная тетрадка: подключение к Impala и загрузка уникальных значений `c_nazn`.

```python
from rail_connectors.connection import connect

# Таблица и параметры выполнения
table_name = 'ods.scd1_z_main_docum'
mem_limit = '8g'

# Подключение (как в 01_07_acq_dash.ipynb)
impala_db = 'sandbox_ai'
impala_queue = 'ai'
impala_user_name = 'Shestopalov-VYur'
impala_keytab_path = '/home/jovyan/test_requests/tech.keytab'
impala_use_credentials = True
impala_update_keytab = True

print(f'table={table_name}, mem_limit={mem_limit}')
```

```python
imp = connect(
    to='IMPALA',
    extra_options={'db': impala_db},
    driver_args={'tez.queue.name': impala_queue},
    kerberos={
        'keytab_path': impala_keytab_path,
        'use_credentials': impala_use_credentials,
        'update_keytab': impala_update_keytab,
    },
    user_params={'user_name': impala_user_name},
)
imp._init_connection()
print('Impala connection initialized')
```

```python
sql_unique_c_nazn = f"""
select distinct coalesce(c_nazn, '') as c_nazn
from {table_name}
"""

with imp:
    imp.execute(f'set MEM_LIMIT={mem_limit}')
    unique_c_nazn_df = imp.fetch(sql_unique_c_nazn)

print(f'rows in unique_c_nazn_df: {len(unique_c_nazn_df):,}')
unique_c_nazn_df
```
