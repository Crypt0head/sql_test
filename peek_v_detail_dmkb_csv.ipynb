```python
from pathlib import Path
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
pd.set_option("display.width", None)

DATA_DIR = Path("/home/jovyan/documents/Equaring/Data")
df = pd.read_csv(
    DATA_DIR / "v_detail_dmkb_202608061323.csv",
    encoding="cp1251",
    sep=";",
)

df.head(5)
```

```python
cols = ["item", "product", "prod_name"]
word = "эквайринг"

for col in cols:
    if col not in df.columns:
        print(f"{col}: колонки нет в файле")
        continue
    mask = df[col].astype(str).str.contains(word, case=False, na=False)
    n = int(mask.sum())
    print(f"{col}: найдено строк = {n}")
    if n:
        display(df.loc[mask, cols].drop_duplicates().head(20))
```
